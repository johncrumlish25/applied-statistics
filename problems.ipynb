```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Problem 1: Exploring Petal Length

```python
# load the dataset and display the first ten rows
data = pd.read_csv("data/iris-data.csv")

data.head(10)
```

```python
# calculate the mean and standard deviation of petal length
mean_petal = data["petal_length"].mean()
std_petal = data["petal_length"].std()

print("Mean:", mean_petal)
print("Standard deviation:", std_petal)
```

```python
# plot a histogram of petal lengths for all 150 flowers
plt.hist(data["petal_length"])

plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.title("Distribution of Petal Length")

plt.show()
```
