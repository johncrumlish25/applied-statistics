```python
# Numerical arrays
import pandas as pd

# Binomial distribution
import numpy as np

# Plotting
import matplotlib.pyplot as plt
```

## Problem 1: Exploring Petal Length

```python
# Load the dataset and display the first ten rows
data = pd.read_csv("data/iris-data.csv")

data.head(10)
```

```python
# Calculate the mean and standard deviation of petal length
mean_petal = data["petal_length"].mean()
std_petal = data["petal_length"].std()

print("Mean:", mean_petal)
print("Standard deviation:", std_petal)
```

The mean and standard deviation of petal length were calculated using the `.mean()` and `.std()`. The mean represents the average petal length, while the standard deviation measures how spread out the petal lengths are around the mean.

These values will be used to create the normal distribution curve.

[Pandas - Descriptive Stats](https://pandas.pydata.org/docs/user_guide/basics.html#descriptive-statistics)

```python
# Plot a histogram of petal lengths for all 150 flowers
plt.hist(data["petal_length"], density=True)

# Create an x range
x = np.linspace(data["petal_length"].min(), data["petal_length"].max(), 100)

# Calculate the normal distribution values
y = (1 / (std_petal * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean_petal) / std_petal) ** 2)

# Plot the normal distribution line
plt.plot(x, y, 'r-')

# Labels
plt.xlabel("Petal Length")
plt.ylabel("Density")
plt.title("Distribution of Petal Length")

plt.show()
```

A histogram was created to show the distribution of petal lengths for all 150 flowers. A normal distribution curve (red line) was added using the calculated mean and standard deviation.

The red line does not follow the histogram particularly closely. This suggests that the petal lengths are not perfectly normally distributed. The histogram shows some differences from the bell-shaped curve expected in a normal distribution.
